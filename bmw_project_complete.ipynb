## 4. Data Cleaning

### 4.1 Standardize text
نزيل المسافات الزائدة ونوحد شكل الكتابة.

> ملاحظة: `str.title()` يحول `i3` إلى `I3`، لذلك سنعيد أسماء BMW i-series لشكلها المعروف بعد التنظيف.

```python
text_cols = ["model", "transmission", "fuelType", "color"]

for col in text_cols:
    df[col] = df[col].astype("string").str.strip()

df["transmission"] = df["transmission"].str.title()
df["fuelType"] = df["fuelType"].str.title()
df["color"] = df["color"].str.title()

# Model: strip first, then normalize known i-series names without damaging Series/X/M naming.
model_map = {
    "i3": "i3", "I3": "i3",
    "i8": "i8", "I8": "i8"
}
df["model"] = df["model"].replace(model_map)

print(df["fuelType"].value_counts(dropna=False))
print()
print(df["transmission"].value_counts(dropna=False))
```

### 4.2 Remove exact duplicates

```python
before = len(df)
df = df.drop_duplicates().copy()
print("Rows removed as exact duplicates:", before - len(df))
```

### 4.3 Convert impossible values to missing values

بدل اختراع قيمة من عندنا، نحول الخطأ الواضح إلى `NaN` ثم نعالجه بطريقة إحصائية مناسبة.

- `price <= 0` غير منطقي كسعر بيع.
- `mpg <= 0` غير منطقي للاستهلاك في هذا التحليل.

```python
df.loc[df["price"] <= 0, "price"] = np
df.loc[df["mpg"] <= 0, "mpg"] = np.nan
```

### 4.4 Remove clearly invalid/extreme records

حسب قواعد المشروع:
- السنوات بعد 2024 تعتبر غير صالحة لهذه العينة.
- Mileage أكبر من أو يساوي 300,000 يُعامل كقيمة متطرفة/خطأ إدخال.
- السعر فوق 150,000 يُراجع كـ extreme outlier في هذه العينة؛ الداتا تحتوي قيمة 350,000 مصممة كخطأ متطرف.

نحذف الصفوف هنا لأننا لا نملك مصدرًا موثوقًا يعطينا القيمة الأصلية الصحيحة.

```python
before = len(df)

df = df[
    (df["year"] <= 2024) &
    (df["mileage"].isna() | (df["mileage"] < 300000)) &
    (df["price"].isna() | (df["price"] <= 150000))
].copy()

print("Rows removed by validity/outlier rules:", before - len(df))
```

### 4.5 Handle Missing Values

لن نحذف كل صف يحتوي NaN لأن ذلك قد يهدر بيانات مفيدة.

- الأعمدة الرقمية: **Median** لأنه أقل تأثرًا بالـoutliers من Mean.
- `color`: **Mode** لأنه categorical.

```python
numeric_missing_cols = ["price", "mileage", "tax", "mpg", "engineSize"]

missing_before = df[numeric_missing_cols + ["color"]].isnull().sum()
print("Before imputation:")
print(missing_before)
```

```python
for col in numeric_missing_cols:
    df[col] = df[col].fillna(df[col].median())

df["color"] = df["color"].fillna(df["color"].mode()[0])

print("Missing values after cleaning:")
print(df.isnull().sum())
```

## 5. Feature Engineering

نضيف Features تساعد التحليل والـDashboard:
- `car_age`
- `price_per_mile`
- `is_luxury_model`
- `mileage_per_year`

استخدام 2024 هنا متسق مع نطاق الداتا وقاعدة السنة في الـproject guide.

```python
df["car_age"] = 2024 - df["year"]

# تجنب القسمة على صفر
df["price_per_mile"] = np.where(
    df["mileage"] > 0,
    df["price"] / df["mileage"],
    np.nan
)

luxury_models = ["M2", "M3", "M4", "M5", "X5", "X6", "X7", "7 Series", "8 Series"]
df["is_luxury_model"] = df["model"].isin(luxury_models)

df["mileage_per_year"] = np.where(
    df["car_age"] > 0,
    df["mileage"] / df["car_age"],
    np.nan
)

df.head()
```

## 8. Exploratory Data Analysis (EDA)

```python
avg_price_by_model = (
    df.groupby("model", as_index=False)["price"]
      .mean()
      .sort_values("price", ascending=False)
)

avg_price_by_model.head(10)
```

```python
year_analysis = (
    df.groupby("year", as_index=False)
      .agg(average_price=("price", "mean"),
           number_of_cars=("listing_id", "count"))
)

year_analysis.tail(10)
```

```python
fuel_analysis = (
    df.groupby("fuelType", as_index=False)
      .agg(
          number_of_cars=("listing_id", "count"),
          average_price=("price", "mean"),
          average_mileage=("mileage", "mean")
      )
      .sort_values("average_price", ascending=False)
)

fuel_analysis
```

```python
transmission_analysis = (
    df.groupby("transmission", as_index=False)
      .agg(
          number_of_cars=("listing_id", "count"),
          average_price=("price", "mean")
      )
      .sort_values("average_price", ascending=False)
)

transmission_analysis
```

```python
correlation = df[["price", "year", "mileage", "engineSize", "mpg", "tax"]].corr()
correlation
```

```python
print("Correlation with price:")
print(correlation["price"].sort_values(ascending=False))
```
